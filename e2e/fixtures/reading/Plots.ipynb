# Plots

Matplotlib figures as PNG and as SVG.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import math
xs = [i / 10 for i in range(100)]
fig, ax = plt.subplots(figsize=(6, 3), dpi=80)
for moon, period in [('Io', 1.769), ('Europa', 3.551), ('Ganymede', 7.155)]:
    ax.plot(xs, [math.sin(2 * math.pi * x / period) for x in xs], label=moon)
ax.set_xlabel('days'); ax.set_ylabel('phase'); ax.legend(); plt.show()
```

```python
from matplotlib_inline.backend_inline import set_matplotlib_formats
set_matplotlib_formats('svg')
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(['Io', 'Europa', 'Ganymede', 'Callisto'], [1821.6, 1560.8, 2634.1, 2410.3])
ax.set_title('Radius (km)'); plt.show()
```

```python
set_matplotlib_formats('png')
fig, ax = plt.subplots(figsize=(12, 2), dpi=80)
ax.plot(xs, [x * x for x in xs]); ax.set_title('A wide figure'); plt.show()
```
