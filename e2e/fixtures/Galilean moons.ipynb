# Galilean moons

The four large moons of Jupiter, found by Galileo in 1610.

```typescript
const moons = [
  { name: "Io", radiusKm: 1821.6 },
  { name: "Europa", radiusKm: 1560.8 },
  { name: "Ganymede", radiusKm: 2634.1 },
  { name: "Callisto", radiusKm: 2410.3 },
];
const largest = moons.reduce((a, b) => (b.radiusKm > a.radiusKm ? b : a));
console.log(`${largest.name} is the largest moon in the solar system`);
```

```typescript
moons.map(m => m.name.toUpperCase()).join(" · ")
```

```typescript
const rows = moons.map(m => `<tr><td>${m.name}</td><td>${m.radiusKm}</td></tr>`).join("");
await jupyter.display(jupyter.html`<table class="moon-table"><tr><th>Moon</th><th>Radius (km)</th></tr>${rows}</table>`);
```
