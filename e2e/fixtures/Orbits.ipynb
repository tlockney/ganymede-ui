# Orbital periods

The Galilean moons, on `ganymede serve`'s TypeScript kernel; $T^2 \propto a^3$.

```typescript
const days = { Io: 1.769, Europa: 3.551, Ganymede: 7.155, Callisto: 16.689 };
for (const [moon, d] of Object.entries(days)) {
  console.log(`${moon.padEnd(9)} ${d.toFixed(3).padStart(6)} days`);
}
```

```typescript
[days.Europa / days.Io, days.Ganymede / days.Europa]
```
